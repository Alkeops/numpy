# Analizando una Imagen

```python
import matplotlib.pyplot as plt
import pandas as pd
img = plt.imread('beagle.jpg')
img.shape

img[10,10]
```

```python
rojo = img[10, 10, 0]
verde = img[10, 10, 1]
azul = img[10, 10, 2]

print("Rojo:", rojo)
print("Verde:", verde)
print("Azul:", azul)
```

```python
print("Dimensiones:", img.shape)
print("Píxel [10,10]:", img[10,10])

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
print("Valor mínimo:", img.min())
print("Valor máximo:", img.max())
print("Media:", img.mean())
print("Desviación estándar:", img.std())
```

```python
plt.imshow(img[:, :, 0], cmap="Reds")
plt.colorbar()
plt.title("Canal Rojo")
plt.show()
```

```python
plt.imshow(img[:, :, 1], cmap="Greens")
plt.colorbar()
plt.title("Canal Verde")
plt.show()
```

```python
plt.imshow(img[:, :, 2], cmap="Blues")
plt.colorbar()
plt.title("Canal Azul")
plt.show()
```

```python
media_rojo = img[:, :, 0].mean()
media_verde = img[:, :, 1].mean()
media_azul = img[:, :, 2].mean()

print("Rojo:", media_rojo)
print("Verde:", media_verde)
print("Azul:", media_azul)
```

```python
canales = ["Rojo", "Verde", "Azul"]
valores = [media_rojo, media_verde, media_azul]

plt.bar(canales, valores)
plt.ylabel("Intensidad promedio")
plt.title("Color promedio de la imagen")
plt.show()
```

```python
datos = img.reshape(-1, 3)

df = pd.DataFrame(
    datos,
    columns=["Rojo", "Verde", "Azul"]
)

print(df.head())
```

```python
plt.hist(img[:,:,2].ravel(), bins=50)
plt.title("Distribución del canal azul")
plt.xlabel("Intensidad")
plt.ylabel("Frecuencia")
plt.show()
```

# Mis ejercicios con la imagen

Una imagen es un arreglo de NumPy de forma `(alto, ancho, 3)`, así que se puede transformar con operaciones de arreglos.

## Escala de grises

Promedio de los tres canales en cada píxel (`axis=2`).

```python
gray = img.mean(axis=2)
print('Forma en grises:', gray.shape)
plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.title('Escala de grises')
plt.show()
```

## Recorte del centro

Slicing con la mitad central de filas y columnas.

```python
height, width = img.shape[:2]
crop = img[height // 4 : 3 * height // 4, width // 4 : 3 * width // 4]
print('Forma del recorte:', crop.shape)
plt.imshow(crop)
plt.axis('off')
plt.title('Centro de la imagen')
plt.show()
```

## Espejo y negativo

`[:, ::-1]` invierte las columnas; `255 - img` invierte los colores.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(img[:, ::-1])
axes[0].set_title('Espejo')
axes[1].imshow(255 - img)
axes[1].set_title('Negativo')
for ax in axes:
    ax.axis('off')
plt.show()
```
