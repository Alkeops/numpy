# Parte 2: Matrices aleatorias

Generar matrices de N×N con Python puro e imprimirlas alineadas.

```python
import random
```

```python
def generar_matriz(N):
    matriz= []
    for i in range (N):
        fila = []
        for j in range(N):
            fila.append(random.randint(1, 9))
        matriz.append(fila)
    return matriz

N = 5

matriz_aleatoria = generar_matriz(N)
for fila in matriz_aleatoria:
    print(fila)

def imprimir_matriz(matriz):
    for fila in matriz:
        fila_formateada = ' '.join(f'{elemento:2}' for elemento in fila)
        print(fila_formateada)
```

```python
print("Matriz de 5x5")
m_5 = generar_matriz(5)
imprimir_matriz(m_5)
print("============")
print("Matriz de 10x10")
m_10 = generar_matriz(10)
imprimir_matriz(m_10)
print("============")
print("Matriz de 20x20")
m_20 = generar_matriz(20)
imprimir_matriz(m_20)
```

# Interactuando con los arreglos

Ejercicios propios para practicar atributos y operaciones de NumPy.

## Tipo de dato, forma y transpuesta

`reshape` cambia la forma sin cambiar los datos y `.T` intercambia filas por columnas.

```python
print('Tipo de dato:', array_1d.dtype)
print('Original', array_1d.shape)
print(array_1d)
print('Reshape a (4, 3)')
print(array_1d.reshape(4, 3))
print('Transpuesta', array_1d.T.shape)
print(array_1d.T)
```

## Sumas por eje

`axis=0` suma hacia abajo (por columna) y `axis=1` suma hacia la derecha (por fila).

```python
grid = np.arange(1, 13).reshape(3, 4)
print(grid)
print('Suma por columna:', grid.sum(axis=0))
print('Suma por fila:', grid.sum(axis=1))
```

## Matriz aleatoria con NumPy

Lo mismo que `generar_matriz`, pero en una sola línea con `np.random.randint`.

```python
random_matrix = np.random.randint(1, 10, size=(5, 5))
print(random_matrix)
print('Promedio:', random_matrix.mean())
print('Máximo por fila:', random_matrix.max(axis=1))
```
